# Further Analysis 

This notebook includes 4 additional interesting pieces of analysis about the data

```python
import sqlite3 as sql
import pandas as pd
from pathlib import Path
from nba.team_dates import plot_team_founding_years
from nba.count_games import count_yearly_games
from nba.largest_stadium import largest_stadium
from nba.height_distribution import height_graph

DATA_PATH = Path.cwd() / 'data' / 'nba.sqlite'
con = sql.connect(DATA_PATH)  # connect to the database
```

```python
#Display the largest Stadium and its Capacity
largest_stadium(con)
```

```python
#This function prints the founding dates for the oldest n NBA teams
plot_team_founding_years(con,5)
```

```python
#Display the number of games played each year
count_yearly_games(con)
```

```python
#Display the Distribution of Heights in the NBA 
height_graph(con)
```
